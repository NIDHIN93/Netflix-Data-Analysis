## USA vs India Netflix content

```python
netflix["date_added"] = pd.to_datetime(netflix["date_added"], errors="coerce")

netflix["year_added"] = netflix["date_added"].dt.year

netflix["first_country"] = netflix["country"].fillna("Unknown").apply(
    lambda x: x.split(",")[0].strip()
)

us_ind = netflix[netflix["first_country"].isin(["United States", "India"])]
```

```python
fig, ax = plt.subplots(1, 1, figsize=(12, 6))

fig.patch.set_facecolor("#141414")
ax.set_facecolor("#141414")

colors = {
    "United States": "#221f1f",  # USA black
    "India": "#E50914"           # India red
}

for country, color in colors.items():
    hs_built = (
        us_ind[us_ind["first_country"] == country]["year_added"]
        .value_counts()
        .sort_index()
    )

    ax.plot(
        hs_built.index,
        hs_built.values,
        color=color,
        linewidth=3,
        label=country
    )

    ax.fill_between(
        hs_built.index,
        0,
        hs_built.values,
        color=color,
        alpha=0.45
    )

ax.set_ylim(0, 1000)
ax.set_xlim(2008, 2020)
plt.xticks(np.arange(2008, 2021, 1))

ax.yaxis.tick_right()
ax.axhline(y=0, color="#F5F5F1", linewidth=1.3, alpha=0.4)

for s in ["top", "right", "bottom", "left"]:
    ax.spines[s].set_visible(False)

ax.grid(False)
ax.margins(x=0)

ax.tick_params(axis="x", colors="#F5F5F1", length=0)
ax.tick_params(axis="y", colors="#F5F5F1", length=0)

fig.text(
    0.5, 0.88,
    "USA vs. India: When was content added?",
    fontsize=16,
    fontweight="bold",
    fontfamily="serif",
    color="#F5F5F1",
    ha="center"
)

fig.text(
    0.5, 0.55,
    """
We know from our work above that Netflix is dominated by the USA & India.
It would also be reasonable to assume that, since Netflix is an American
company, Netflix increased content first in the USA, before other nations.

That is exactly what we see here: a slow and then rapid increase in content
for the USA, followed by Netflix being launched to the Indian market in 2016.
""",
    fontsize=12,
    fontweight="light",
    fontfamily="serif",
    color="#F5F5F1",
    ha="center",
    va="center"
)

fig.text(0.43, 0.15, "India", fontweight="bold", fontfamily="serif", fontsize=15, color="#E50914")
fig.text(0.49, 0.15, "|", fontweight="bold", fontfamily="serif", fontsize=15, color="#F5F5F1")
fig.text(0.51, 0.15, "USA", fontweight="bold", fontfamily="serif", fontsize=15, color="#221f1f")

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib
from PIL import Image, ImageDraw
from wordcloud import WordCloud

# Text from Netflix titles
text = " ".join(netflix["title"].dropna().astype(str))

# Netflix color map
cmap = matplotlib.colors.LinearSegmentedColormap.from_list(
    "netflix_cmap",
    ["#221f1f", "#b20710", "#E50914"]
)

# Create Netflix N mask
width, height = 800, 800
mask_img = Image.new("L", (width, height), 255)
draw = ImageDraw.Draw(mask_img)

# N shape sizes
left_x1, left_x2 = 250, 340
right_x1, right_x2 = 460, 550
top_y, bottom_y = 180, 620

# Left vertical bar
draw.rectangle([left_x1, top_y, left_x2, bottom_y], fill=0)

# Right vertical bar
draw.rectangle([right_x1, top_y, right_x2, bottom_y], fill=0)

# Diagonal middle bar
draw.polygon(
    [
        (left_x2, top_y),
        (right_x2, bottom_y),
        (right_x1, bottom_y),
        (left_x1, top_y)
    ],
    fill=0
)

mask = np.array(mask_img)

# Create word cloud
wordcloud = WordCloud(
    background_color="white",
    width=800,
    height=800,
    colormap=cmap,
    max_words=300,
    mask=mask,
    contour_width=0,
    random_state=42
).generate(text)

# Show output
plt.figure(figsize=(8, 8))
plt.imshow(wordcloud, interpolation="bilinear")
plt.axis("off")
plt.tight_layout(pad=0)
plt.show()
```
